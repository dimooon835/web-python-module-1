## Учёт заявок в IT-поддержке”

В компании есть небольшая IT-поддержка. Сотрудники отправляют заявки: `не работает принтер`, `забыл пароль`, `нет интернета`. Нужно сделать консольную программу, которая помогает хранить и обрабатывать эти заявки.

### Что должна уметь программа

Меню:
1. Добавить заявку
2. Показать все заявки
3. Изменить статус заявки
4. Показать открытые заявки
5. Найти заявку по имени сотрудника
0. Выход

### Данные одной заявки
```txt
Номер заявки
Имя сотрудника
Описание проблемы
Статус
```

**Статусы можно сделать такими:**
`новая`, `в работе`, `закрыта`


### Пример заявки
```txt
№1
Сотрудник: Иван
Проблема: Не работает принтер
Статус: новая
```

### Обязательные условия
1. Номер заявки создаётся автоматически.
2. При добавлении заявки статус всегда `новая`.
3. Можно изменить статус только на:
    - `новая`
    - `в работе`
    - `закрыта`
4. При поиске по имени программа должна показывать все заявки этого сотрудника.
5. Открытые заявки — это заявки со статусом `новая` или `в работе`.
6. Решение должно быть реализовано в объектно-ориентированном стиле с соблюдением принципов `SOLID`:
   - отдельный класс для заявки;
   - отдельный класс или сервис для управления заявками;
   - отдельный класс для работы с консольным меню;
   - классы не должны брать на себя лишние обязанности;
   - зависимости между частями программы должны быть минимальными и понятными.

````python
class Ticket:
    def __init__(self, ticket_id, employee_name, description):
        self.ticket_id = ticket_id,
        self.employee_name = employee_name,
        self.description = description,
        self.status = "новая"

    def __str__(self):
        return(f"№ {self.ticket_id}\n"
               f"Сотрудник: {self.employee_name}\n"
               f"Проблема: {self.description}\n"
               f"Статус: {self.status}")


class TicketService:
    def __init__(self):
        self.tickets = []
        self.next_id = 0

    def add_ticket(self, employee_name, description):
        ticket = Ticket(self.next_id, employee_name, description)
        self.tickets.append(ticket)
        self.next_id += 1
        return ticket
    
    def get_all_tickets(self):
        return self.tickets
    
    def change_status(self, ticket_id, new_status):
        allowed_statuses = {"новая" , "в работе", "закрыта"}
        
        if new_status not in allowed_statuses:
            return False
        
        for ticket in self.tickets:
            if ticket.ticket_id == ticket_id:
                ticket.stasus = new_status
                return True
        return False
    
    def get_open_tickets(self):
        return [t for t in self.tickets if t.status in ("новая", "в работе")]
    
    def find_by_employee(self, name):
        return [t for t in self.tickets if t.employee_name.lower() == name.lower()]
    

class Console:
    def __init__(self):
        self.service = TicketService()

    def run(self):
        while True:
            print("\nМеню.")
            print("1. Добавить заявку.")
            print("2. Показать все заявки.")
            print("3. Изменить статус заявки.")
            print("4. Показать открытые заявки.")
            print("5. Найти заявку по имени сотрудника.")
            print("0. Выход.")

            choice = input("Выберите пункт: ")

            if choice == "1":
                self.add_ticket()
            elif choice == "2":
                self.show_all_tickets()
            elif choice == "3":
                self.change_status()
            elif choice == "4":
                self.show_open_tickets()
            elif choice == "5":
                self.find_by_employee()
            elif choice == "0":
                break
            else:
                print("Неверный пункт меню.")

    def add_ticket(self):
        name = input("Имя сотрудника: ")
        desc = input("Описание проблемы: ")
        self.service.add_ticket(name, desc)
        print("Заявка добавлена")

    def show_all_tickets(self):
        tickets = self.service.get_all_tickets()

        if not tickets:
            print("Заявки отсутствуют.")
            return
        
        for t in tickets:
            print(t)
            print("-" * 20)

    def change_status(self):
        try:
            ticket_id = int(input("Номер заявки: "))
            new_status = input("Новый статус: ")

            if self.service.change_status(ticket_id, new_status):
                print("Статус изменён.")
            else:
                print("Ошибка.")

        except ValueError:
            print("Неверный номер заявки.")

    def show_open_tickets(self):
        tickets = self.service.get_open_tickets()

        if not tickets:
            print("Открытых заявок нет.")
            return
        
        for t in tickets:
            print(t)
            print("-" * 20)

    def find_by_employee(self):
        name = input("Имя сотрудника: ")
        tickets = self.service.find_by_employee()

        if not tickets:
            print("Заявки не найдены.")
            return
        
        for t in tickets:
            print(t)
            print("-" * 20)

if __name__ == "__main__":
    menu = Console()
    menu.run()
````
