## Учёт заявок в IT-поддержке”

В компании есть небольшая IT-поддержка. Сотрудники отправляют заявки: `не работает принтер`, `забыл пароль`, `нет интернета`. Нужно сделать консольную программу, которая помогает хранить и обрабатывать эти заявки.

### Что должна уметь программа

Меню:
1. Добавить заявку
2. Показать все заявки
3. Изменить статус заявки
4. Показать открытые заявки
5. Найти заявку по имени сотрудника
0. Выход

### Данные одной заявки
```txt
Номер заявки
Имя сотрудника
Описание проблемы
Статус
```

**Статусы можно сделать такими:**
`новая`, `в работе`, `закрыта`


### Пример заявки
```txt
№1
Сотрудник: Иван
Проблема: Не работает принтер
Статус: новая
```

### Обязательные условия
1. Номер заявки создаётся автоматически.
2. При добавлении заявки статус всегда `новая`.
3. Можно изменить статус только на:
    - `новая`
    - `в работе`
    - `закрыта`
4. При поиске по имени программа должна показывать все заявки этого сотрудника.
5. Открытые заявки — это заявки со статусом `новая` или `в работе`.
6. Решение должно быть реализовано в объектно-ориентированном стиле с соблюдением принципов `SOLID`:
   - отдельный класс для заявки;
   - отдельный класс или сервис для управления заявками;
   - отдельный класс для работы с консольным меню;
   - классы не должны брать на себя лишние обязанности;
   - зависимости между частями программы должны быть минимальными и понятными.

````python
from typing import List


class Ticket:
    _id_counter = 1

    VALID_STATUSES = ["новая", "в работе", "закрыта"]

    def __init__(self, employee_name: str, description: str):
        self.id = Ticket._id_counter
        Ticket._id_counter += 1

        self.employee_name = employee_name
        self.description = description
        self.status = "новая"

    def set_status(self, new_status: str):
        if new_status not in Ticket.VALID_STATUSES:
            raise ValueError("Недопустимый статус")
        self.status = new_status

    def __str__(self):
        return (f"№{self.id}\n"
                f"Сотрудник: {self.employee_name}\n"
                f"Проблема: {self.description}\n"
                f"Статус: {self.status}\n")


class TicketService:
    def __init__(self):
        self.tickets: List[Ticket] = []

    def add_ticket(self, employee_name: str, description: str):
        ticket = Ticket(employee_name, description)
        self.tickets.append(ticket)

    def get_all_tickets(self) -> List[Ticket]:
        return self.tickets

    def find_by_employee(self, name: str) -> List[Ticket]:
        return [t for t in self.tickets if t.employee_name.lower() == name.lower()]

    def get_open_tickets(self) -> List[Ticket]:
        return [t for t in self.tickets if t.status in ["новая", "в работе"]]

    def update_status(self, ticket_id: int, new_status: str) -> bool:
        for ticket in self.tickets:
            if ticket.id == ticket_id:
                ticket.set_status(new_status)
                return True
        return False


class ConsoleUI:
    def __init__(self, service: TicketService):
        self.service = service

    def run(self):
        while True:
            self.print_menu()
            choice = input("Выберите пункт: ")

            if choice == "1":
                self.add_ticket()
            elif choice == "2":
                self.show_all()
            elif choice == "3":
                self.update_status()
            elif choice == "4":
                self.show_open()
            elif choice == "5":
                self.search_by_employee()
            elif choice == "0":
                print("Выход...")
                break
            else:
                print("Неверный выбор")

    def print_menu(self):
        print("\n--- Меню ---")
        print("1. Добавить заявку")
        print("2. Показать все заявки")
        print("3. Изменить статус заявки")
        print("4. Показать открытые заявки")
        print("5. Найти заявку по имени сотрудника")
        print("0. Выход")

    def add_ticket(self):
        name = input("Имя сотрудника: ")
        description = input("Описание проблемы: ")
        self.service.add_ticket(name, description)
        print("Заявка добавлена")

    def show_all(self):
        tickets = self.service.get_all_tickets()
        if not tickets:
            print("Нет заявок")
            return
        for t in tickets:
            print(t)

    def update_status(self):
        try:
            ticket_id = int(input("Введите номер заявки: "))
            print("Доступные статусы:", ", ".join(Ticket.VALID_STATUSES))
            new_status = input("Введите новый статус: ")
            if self.service.update_status(ticket_id, new_status):
                print("Статус обновлён")
            else:
                print("Заявка не найдена")
        except ValueError:
            print("Ошибка ввода")

    def show_open(self):
        tickets = self.service.get_open_tickets()
        if not tickets:
            print("Нет открытых заявок")
            return
        for t in tickets:
            print(t)

    def search_by_employee(self):
        name = input("Введите имя сотрудника: ")
        tickets = self.service.find_by_employee(name)
        if not tickets:
            print("Заявки не найдены")
            return
        for t in tickets:
            print(t)


if __name__ == "__main__":
    service = TicketService()
    ui = ConsoleUI(service)
    ui.run()
````
